```python
class library:
    def __init__(self):
        self.Books = []
    def add_book(self,book_name):
        self.Books.append(book_name)
        print(book_name,"has been added")
    def view_books(self):
        if len(self.Books)==0:
            print("books are not availabel")
        else:
            ("books are availabel")
        for i in self.Books:
            print(i)
    def rem_book(self,book_name):
        if book_name in self.Books:
            self.Books.remove(book_name)
            print(book_name,"book has removed")
        else:
            print("book is not availabe")
```

```python
stu=library()
```
